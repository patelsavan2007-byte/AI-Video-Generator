```python
import torch
from diffusers import DiffusionPipeline

# switch to "mps" for apple devices
pipe = DiffusionPipeline.from_pretrained("SG161222/RealVisXL_V5.0", dtype=torch.bfloat16, device_map="cuda")
```

```python
prompt = '''
cinematic realistic portrait of a young indian man,
short messy black hair,
brown eyes,
light brown skin,
dark oversized hoodie,
neutral expression,
front view,
studio lighting,
high detail,
sharp focus,
consistent facial structure
'''
```

```python
image = pipe(prompt,
    negative_prompt="blurry, low quality, distorted",
    height=512,
    width=512,
    num_inference_steps=30,
    guidance_scale=7.5,
    num_images_per_prompt=1).images[0]
```

```python
image.save("test.png")
```
